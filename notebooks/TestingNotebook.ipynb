```julia
using CSV
using DataFrames
using BehaviorModels
using StatsBase
using Turing
```

```julia
	data_path = "../data"
	file_name = "processed_rat_data.csv.gz"

	# load the data
	rat_df = CSV.read(joinpath(data_path, file_name), DataFrame)

	# useful variables for later
	rats = unique(rat_df.name);
```

```julia
# calculate average number of trials per rat
n_trials = combine(groupby(rat_df, :name), nrow => :n_trials)
```

```julia
	pacf_dict = Dict()
	for rat in rats
		rts = rat_df[rat_df.name .== rat, :rt]
		pacf_dict[rat] = pacf(log.(rts), 1:100)
	end
	pacf_df = DataFrame(pacf_dict)
```

```julia
	rat_behavior = rat_df[rat_df.name .== "Daenerys", :]
	behavior_trials = [BehaviorTrial(r.delta_flashes, r.choose_right, r.correct, r.rt) for r in eachrow(rat_behavior)]
	
	model = simple_ar_model(behavior_trials, 6)
	posterior_sample = sample(model, NUTS(), 200)
```
